```python
%%writefile panen.py

def hitung_total_panen(jumlah_kg, harga_per_kg):
    return jumlah_kg * harga_per_kg


def hitung_diskon(total, persen_diskon):
    return total - (total * persen_diskon / 100)


jumlah = 500
harga = 7000

total = hitung_total_panen(jumlah, harga)
total_setelah_diskon = hitung_diskon(total, 10)

print("Jumlah panen :", jumlah, "kg")
print("Harga per kg :", harga)
print("Total hasil panen : Rp", total)
print("Total setelah diskon 10% : Rp", total_setelah_diskon)
```
